## Part 7: Summarizing with `groupby()`

## Step 1: Build a word frequency table

### Demonstration


```
sentence = "numpy makes numerical computing fast and numpy makes arrays powerful and pandas makes tables powerful"

tokens = sentence.lower().split()
tokens_array = np.array(tokens)
vocabulary = np.unique(tokens_array)

counts = [np.sum(tokens_array == word) for word in vocabulary]

word_freq = pd.DataFrame({'word': vocabulary, 'count': counts})
word_freq = word_freq.sort_values('count', ascending=False).reset_index(drop=True)
word_freq
```

*Your answer:*

Comparing the pandas approach to raw NumPy for word frequency analysis reveals clear advantages and trade-offs:

**What pandas makes easier:**
*   **Readability and Intuition:** Column names (`'word'`, `'count'`, `'sentence_a_count'`) make the code much more self-explanatory and easier to understand at a glance, compared to remembering what each index in a NumPy array represents.
*   **Direct Operations on Labeled Data:** Filtering (`table[table['count'] == 1]`), sorting (`word_freq.sort_values()`), and creating new columns (`table['difference'] = ...`) are straightforward and intuitive with pandas, operating directly on the meaningful labels.
*   **Data Aggregation:** While not explicitly shown in this mini-project, pandas' `groupby()` and `pivot_table()` functions are incredibly powerful for complex aggregations that would be very tedious and error-prone to implement with raw NumPy loops.
*   **Handling Mixed Data Types:** DataFrames can easily handle columns with different data types (strings for words, integers for counts), which is a key advantage over NumPy arrays that typically require a single data type.

**What pandas hides (that NumPy made explicit):**
*   **Underlying Array Operations:** Pandas abstracts away the direct array manipulations. When you perform an operation on a Series or DataFrame, you don't directly see the NumPy array operations happening underneath. This can be good for simplicity but might hide performance details if you're working with extremely large datasets and need fine-grained control.
*   **Memory Layout:** NumPy's explicit nature means you are more aware of the contiguous memory blocks and how data is stored, which can be important for highly optimized algorithms. Pandas handles this automatically.

In essence, pandas provides a higher-level, more semantic API that prioritizes ease of use and expressiveness for tabular data analysis, building on the numerical efficiency of NumPy arrays underneath.

## Final Reflection Questions

1.  **How is a pandas DataFrame related to a NumPy array? (Think back to last week's lab.)**
    A pandas DataFrame is built on top of NumPy arrays. Internally, each column in a DataFrame is typically stored as a NumPy array. This means that DataFrames inherit the performance benefits of NumPy's vectorized operations. While a NumPy array is a homogeneous collection of elements, a DataFrame can hold multiple NumPy arrays (one for each column) and provides additional features like labeled rows (index) and columns, handling of mixed data types across columns, and powerful methods for data manipulation that go beyond what raw NumPy offers.

2.  **Name one pandas operation that would be painful to do with raw NumPy arrays and loops instead, and explain why.**
    The `groupby()` operation, especially with multiple aggregations (`.agg()`), would be particularly painful with raw NumPy arrays and loops. If you wanted to calculate the mean `total_bill` and mean `tip` for each `day` using NumPy, you would have to:
    1.  Manually identify the unique days.
    2.  Loop through each unique day.
    3.  For each day, filter the NumPy array to get only the rows corresponding to that day.
    4.  Extract the `total_bill` and `tip` values for those filtered rows.
    5.  Calculate the mean of `total_bill` and `tip` for that subset.
    6.  Store these results in a new structure.
    
    This process involves several manual filtering steps, nested loops, and explicit management of the resulting aggregated data, which is highly error-prone and much more verbose compared to the single line `tips.groupby('day').agg(avg_bill=('total_bill','mean'), avg_tip=('tip','mean'))` in pandas.
