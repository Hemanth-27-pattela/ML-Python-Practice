## Part 1 — Distance Measures

KNN's entire decision rule is "find the closest points." "Closest" needs a precise definition — that's what a distance metric gives us.

### 1.1 Demo: Euclidean and Manhattan distance, by hand


```
p1 = np.array([5.1, 3.5, 1.4, 0.2])   # a sample flower's 4 measurements
p2 = np.array([6.4, 3.2, 4.5, 1.5])   # a different flower

euclidean = np.sqrt(np.sum((p1 - p2) ** 2))
manhattan = np.sum(np.abs(p1 - p2))

print('Euclidean distance:', euclidean)
print('Manhattan distance:', manhattan)
```

**Predict, then think:** Euclidean distance is always ≤ or ≥ Manhattan distance for the same two points (pick one) — why? Think about what happens along a single axis vs. across multiple axes at once.

*Your answer:* Euclidean distance is always **≤** Manhattan distance.

This is because of the triangle inequality. Manhattan distance sums the absolute differences along each axis, effectively moving in a grid-like pattern. Euclidean distance takes the straight-line path (hypotenuse) between two points. The sum of the lengths of the two shorter sides of a right triangle (Manhattan-like movement along axes) is always greater than or equal to the length of the hypotenuse (Euclidean distance). The equality holds only when movement is purely along one axis (i.e., points differ in only one dimension).

### Student Practice — Minkowski distance

The Minkowski distance generalizes both: $d(p,q) = \left(\sum |p_i - q_i|^p\right)^{1/p}$. Euclidean is Minkowski with $p=2$; Manhattan is Minkowski with $p=1$.

Write a function `minkowski(p1, p2, p_order)` that computes this, then confirm it reproduces the Euclidean and Manhattan values above when `p_order=2` and `p_order=1`.


```
def minkowski(p1, p2, p_order):
    return np.sum(np.abs(p1 - p2)**p_order)**(1/p_order)

# Test with p_order = 1 (Manhattan)
manhattan_check = minkowski(p1, p2, 1)
print(f'Minkowski (p=1) / Manhattan distance: {manhattan_check}')

# Test with p_order = 2 (Euclidean)
euclidean_check = minkowski(p1, p2, 2)
print(f'Minkowski (p=2) / Euclidean distance: {euclidean_check}')

# Original values for comparison
euclidean = np.sqrt(np.sum((p1 - p2) ** 2))
manhattan = np.sum(np.abs(p1 - p2))
print(f'Original Euclidean distance: {euclidean}')
print(f'Original Manhattan distance: {manhattan}')
```

## Part 2 — Fitting a KNN Classifier

Same rule from Week 5 applies here: **split first, then scale using train statistics only.** KNN is a distance-based method, so unscaled features (e.g. one column in the 1000s, another in single digits) would silently dominate every distance calculation — scaling matters *more* here than it did for linear regression.


```
X = df[iris.feature_names]
y = df['target']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=1, stratify=y)

scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```
knn5 = KNeighborsClassifier(n_neighbors=5)
knn5.fit(X_train_scaled, y_train)

y_pred = knn5.predict(X_test_scaled)
print('Accuracy (k=5):', accuracy_score(y_test, y_pred))

ConfusionMatrixDisplay(confusion_matrix(y_test, y_pred), display_labels=iris.target_names).plot()
plt.show()
```

## Part 3 — Why K Matters

### Demo: sweep K from 1 to 30


```
train_acc, test_acc = [], []
k_values = range(1, 31)

for k in k_values:
    knn = KNeighborsClassifier(n_neighbors=k)
    knn.fit(X_train_scaled, y_train)
    train_acc.append(accuracy_score(y_train, knn.predict(X_train_scaled)))
    test_acc.append(accuracy_score(y_test, knn.predict(X_test_scaled)))

plt.figure(figsize=(8, 4))
plt.plot(list(k_values), train_acc, label='train accuracy')
plt.plot(list(k_values), test_acc, label='test accuracy')
plt.xlabel('K (number of neighbors)')
plt.ylabel('accuracy')
plt.legend()
plt.title('Effect of K on train vs test accuracy')
plt.show()
```

**Reflection:** This should look familiar from Week 6. At K=1, is the model closer to overfitting or underfitting? What about at K=30? Connect this explicitly to bias and variance.

*Your answer:*
-   **At K=1**: The model is closer to **overfitting**. A K=1 model is highly sensitive to noise in the training data because it makes predictions based solely on the single closest neighbor. This leads to very low bias (it fits the training data perfectly, or nearly so) but very high variance (it's unstable and changes significantly with slight variations in the training data, performing poorly on unseen data).
-   **At K=30**: The model is closer to **underfitting**. A very large K means that the model considers many neighbors, averaging out local patterns. This results in high bias (it might miss important local structures and simplify the decision boundary too much) but lower variance (it's more stable and less affected by individual noisy data points, but might not capture the true underlying relationships).

```
best_k_index = np.argmax(test_acc)
best_k = k_values[best_k_index]
highest_test_accuracy = test_acc[best_k_index]

print(f"The K with the highest test accuracy is: {best_k}")
print(f"Highest test accuracy: {highest_test_accuracy:.4f}")

# If there are multiple K values with the same highest accuracy, find them all
all_best_k_indices = np.where(np.array(test_acc) == highest_test_accuracy)[0]
all_best_k_values = [k_values[i] for i in all_best_k_indices]

if len(all_best_k_values) > 1:
    print(f"Other K values with the same highest test accuracy: {all_best_k_values}")
    print("To pick among ties, it's generally safer to choose a slightly higher K to reduce variance (overfitting) if the accuracy is comparable, as very low K can be highly sensitive to noise in new data. For example, if both K=3 and K=7 give the same highest accuracy, K=7 might be preferred for better generalization.")
```

### Hand-worked solution for Query point Q = (5, 5)

1.  **Calculate Euclidean Distances:**
    -   **Q to A (2, 4):** $d = \sqrt{(5-2)^2 + (5-4)^2} = \sqrt{3^2 + 1^2} = \sqrt{9 + 1} = \sqrt{10} \approx 3.162$
    -   **Q to B (4, 6):** $d = \sqrt{(5-4)^2 + (5-6)^2} = \sqrt{1^2 + (-1)^2} = \sqrt{1 + 1} = \sqrt{2} \approx 1.414$
    -   **Q to C (4, 2):** $d = \sqrt{(5-4)^2 + (5-2)^2} = \sqrt{1^2 + 3^2} = \sqrt{1 + 9} = \sqrt{10} \approx 3.162$
    -   **Q to D (6, 4):** $d = \sqrt{(5-6)^2 + (5-4)^2} = \sqrt{(-1)^2 + 1^2} = \sqrt{1 + 1} = \sqrt{2} \approx 1.414$
    -   **Q to E (6, 2):** $d = \sqrt{(5-6)^2 + (5-2)^2} = \sqrt{(-1)^2 + 3^2} = \sqrt{1 + 9} = \sqrt{10} \approx 3.162$

2.  **List the 3 nearest points and their classes (for K=3):**
    -   B (distance $\approx$ 1.414, Class: Red)
    -   D (distance $\approx$ 1.414, Class: Blue)
    -   (Tie between A, C, E, all $\approx$ 3.162. Let's pick C for now to demonstrate, but in a real scenario, tie-breaking rules would apply, or one might reconsider K or the metric.)
    -   *Correction based on code's behavior: If distances are identical, order in input might break ties. For strict K=3, we need three distinct closest. Let's take B, D, and C (or A or E) as the next closest for demonstration.*
    -   **Nearest 3 (sorted by distance):**
        -   B (distance $\approx$ 1.414, Class: Red)
        -   D (distance $\approx$ 1.414, Class: Blue)
        -   A, C, E (all $\approx$ 3.162). If we need *exactly* three, we'd need a tie-breaking rule. Assuming alphabetical or input order for the third, let's say A.
        -   So, B (Red), D (Blue), A (Red).

3.  **Majority Vote:**
    -   Red: 2 votes (from B, A)
    -   Blue: 1 vote (from D)
    -   **Predicted Class for Q: Red**


```
points = {'A': (2, 4, 'Red'), 'B': (4, 6, 'Red'), 'C': (4, 2, 'Blue'), 'D': (6, 4, 'Blue'), 'E': (6, 2, 'Blue')}
Q = np.array([5, 5])

distances = {}
for name, (x, y_, label) in points.items():
    d = np.sqrt((Q[0] - x) ** 2 + (Q[1] - y_) ** 2)
    distances[name] = (round(d, 3), label)

for name, (d, label) in sorted(distances.items(), key=lambda item: item[1][0]):
    print(f'{name}: distance={d}, class={label}')
```
